```python
# ============================================================
# STUDENT & CAMPUS UTILITY MANAGEMENT SYSTEM
# Python + SQLite
# ============================================================

import sqlite3
from datetime import datetime

# ------------------------------------------------------------
# DATABASE CONNECTION
# ------------------------------------------------------------

conn = sqlite3.connect("campus.db")
cursor = conn.cursor()


# ------------------------------------------------------------
# CREATE DATABASE TABLES
# ------------------------------------------------------------

cursor.execute("""
CREATE TABLE IF NOT EXISTS students (
    student_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    department TEXT NOT NULL,
    year TEXT NOT NULL,
    phone TEXT,
    email TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS attendance (
    attendance_id INTEGER PRIMARY KEY AUTOINCREMENT,
    student_id INTEGER,
    date TEXT,
    status TEXT,
    FOREIGN KEY(student_id) REFERENCES students(student_id)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS notices (
    notice_id INTEGER PRIMARY KEY AUTOINCREMENT,
    title TEXT NOT NULL,
    description TEXT,
    date TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS complaints (
    complaint_id INTEGER PRIMARY KEY AUTOINCREMENT,
    student_id INTEGER,
    category TEXT,
    description TEXT,
    status TEXT,
    FOREIGN KEY(student_id) REFERENCES students(student_id)
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS facilities (
    facility_id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    description TEXT
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS events (
    event_id INTEGER PRIMARY KEY AUTOINCREMENT,
    event_name TEXT,
    date TEXT,
    venue TEXT,
    description TEXT
)
""")

conn.commit()


# ------------------------------------------------------------
# STUDENT MANAGEMENT
# ------------------------------------------------------------

def add_student():
    print("\n------ ADD STUDENT ------")

    name = input("Enter student name: ")
    department = input("Enter department: ")
    year = input("Enter year: ")
    phone = input("Enter phone number: ")
    email = input("Enter email: ")

    cursor.execute("""
    INSERT INTO students
    (name, department, year, phone, email)
    VALUES (?, ?, ?, ?, ?)
    """, (name, department, year, phone, email))

    conn.commit()

    print("\nStudent added successfully!")


def view_students():
    print("\n------ STUDENT LIST ------")

    cursor.execute("SELECT * FROM students")
    students = cursor.fetchall()

    if not students:
        print("No students found.")
        return

    for student in students:
        print("--------------------------------")
        print("Student ID :", student[0])
        print("Name       :", student[1])
        print("Department :", student[2])
        print("Year       :", student[3])
        print("Phone      :", student[4])
        print("Email      :", student[5])


def search_student():
    print("\n------ SEARCH STUDENT ------")

    student_id = input("Enter student ID: ")

    cursor.execute(
        "SELECT * FROM students WHERE student_id = ?",
        (student_id,)
    )

    student = cursor.fetchone()

    if student:
        print("\nStudent Found!")
        print("ID         :", student[0])
        print("Name       :", student[1])
        print("Department :", student[2])
        print("Year       :", student[3])
        print("Phone      :", student[4])
        print("Email      :", student[5])
    else:
        print("Student not found.")


def update_student():
    print("\n------ UPDATE STUDENT ------")

    student_id = input("Enter student ID: ")

    cursor.execute(
        "SELECT * FROM students WHERE student_id = ?",
        (student_id,)
    )

    student = cursor.fetchone()

    if not student:
        print("Student not found.")
        return

    name = input("Enter new name: ")
    department = input("Enter new department: ")
    year = input("Enter new year: ")
    phone = input("Enter new phone: ")
    email = input("Enter new email: ")

    cursor.execute("""
    UPDATE students
    SET name = ?, department = ?, year = ?, phone = ?, email = ?
    WHERE student_id = ?
    """, (name, department, year, phone, email, student_id))

    conn.commit()

    print("Student updated successfully!")


def delete_student():
    print("\n------ DELETE STUDENT ------")

    student_id = input("Enter student ID: ")

    cursor.execute(
        "SELECT * FROM students WHERE student_id = ?",
        (student_id,)
    )

    student = cursor.fetchone()

    if not student:
        print("Student not found.")
        return

    cursor.execute(
        "DELETE FROM students WHERE student_id = ?",
        (student_id,)
    )

    conn.commit()

    print("Student deleted successfully!")


def student_management():
    while True:
        print("\n================================")
        print("       STUDENT MANAGEMENT")
        print("================================")
        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Update Student")
        print("5. Delete Student")
        print("6. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_student()

        elif choice == "2":
            view_students()

        elif choice == "3":
            search_student()

        elif choice == "4":
            update_student()

        elif choice == "5":
            delete_student()

        elif choice == "6":
            break

        else:
            print("Invalid choice.")


# ------------------------------------------------------------
# ATTENDANCE MANAGEMENT
# ------------------------------------------------------------

def mark_attendance():
    print("\n------ MARK ATTENDANCE ------")

    student_id = input("Enter student ID: ")

    cursor.execute(
        "SELECT * FROM students WHERE student_id = ?",
        (student_id,)
    )

    student = cursor.fetchone()

    if not student:
        print("Student not found.")
        return

    status = input("Enter attendance (Present/Absent): ")

    date = datetime.now().strftime("%Y-%m-%d")

    cursor.execute("""
    INSERT INTO attendance
    (student_id, date, status)
    VALUES (?, ?, ?)
    """, (student_id, date, status))

    conn.commit()

    print("Attendance recorded successfully!")


def view_attendance():
    print("\n------ ATTENDANCE ------")

    student_id = input("Enter student ID: ")

    cursor.execute("""
    SELECT attendance_id, date, status
    FROM attendance
    WHERE student_id = ?
    """, (student_id,))

    records = cursor.fetchall()

    if not records:
        print("No attendance records found.")
        return

    for record in records:
        print("--------------------------------")
        print("Record ID :", record[0])
        print("Date      :", record[1])
        print("Status    :", record[2])


def attendance_management():
    while True:
        print("\n================================")
        print("       ATTENDANCE MANAGEMENT")
        print("================================")
        print("1. Mark Attendance")
        print("2. View Attendance")
        print("3. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            mark_attendance()

        elif choice == "2":
            view_attendance()

        elif choice == "3":
            break

        else:
            print("Invalid choice.")


# ------------------------------------------------------------
# NOTICE MANAGEMENT
# ------------------------------------------------------------

def add_notice():
    print("\n------ ADD NOTICE ------")

    title = input("Enter notice title: ")
    description = input("Enter notice description: ")

    date = datetime.now().strftime("%Y-%m-%d")

    cursor.execute("""
    INSERT INTO notices
    (title, description, date)
    VALUES (?, ?, ?)
    """, (title, description, date))

    conn.commit()

    print("Notice added successfully!")


def view_notices():
    print("\n------ CAMPUS NOTICES ------")

    cursor.execute("SELECT * FROM notices")
    notices = cursor.fetchall()

    if not notices:
        print("No notices available.")
        return

    for notice in notices:
        print("--------------------------------")
        print("Notice ID   :", notice[0])
        print("Title       :", notice[1])
        print("Description :", notice[2])
        print("Date        :", notice[3])


def notice_management():
    while True:
        print("\n================================")
        print("        CAMPUS NOTICES")
        print("================================")
        print("1. Add Notice")
        print("2. View Notices")
        print("3. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_notice()

        elif choice == "2":
            view_notices()

        elif choice == "3":
            break

        else:
            print("Invalid choice.")


# ------------------------------------------------------------
# COMPLAINT MANAGEMENT
# ------------------------------------------------------------

def add_complaint():
    print("\n------ REGISTER COMPLAINT ------")

    student_id = input("Enter student ID: ")

    cursor.execute(
        "SELECT * FROM students WHERE student_id = ?",
        (student_id,)
    )

    student = cursor.fetchone()

    if not student:
        print("Student not found.")
        return

    category = input("Enter complaint category: ")
    description = input("Enter complaint description: ")

    status = "Pending"

    cursor.execute("""
    INSERT INTO complaints
    (student_id, category, description, status)
    VALUES (?, ?, ?, ?)
    """, (student_id, category, description, status))

    conn.commit()

    print("Complaint registered successfully!")


def view_complaints():
    print("\n------ COMPLAINTS ------")

    cursor.execute("""
    SELECT complaint_id, student_id, category,
           description, status
    FROM complaints
    """)

    complaints = cursor.fetchall()

    if not complaints:
        print("No complaints found.")
        return

    for complaint in complaints:
        print("--------------------------------")
        print("Complaint ID :", complaint[0])
        print("Student ID   :", complaint[1])
        print("Category     :", complaint[2])
        print("Description  :", complaint[3])
        print("Status       :", complaint[4])


def update_complaint():
    print("\n------ UPDATE COMPLAINT ------")

    complaint_id = input("Enter complaint ID: ")

    status = input(
        "Enter new status (Pending/Resolved): "
    )

    cursor.execute("""
    UPDATE complaints
    SET status = ?
    WHERE complaint_id = ?
    """, (status, complaint_id))

    conn.commit()

    print("Complaint status updated!")


def complaint_management():
    while True:
        print("\n================================")
        print("       COMPLAINT MANAGEMENT")
        print("================================")
        print("1. Register Complaint")
        print("2. View Complaints")
        print("3. Update Complaint")
        print("4. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_complaint()

        elif choice == "2":
            view_complaints()

        elif choice == "3":
            update_complaint()

        elif choice == "4":
            break

        else:
            print("Invalid choice.")


# ------------------------------------------------------------
# CAMPUS FACILITIES
# ------------------------------------------------------------

def add_facility():
    print("\n------ ADD FACILITY ------")

    name = input("Enter facility name: ")
    description = input("Enter facility description: ")

    cursor.execute("""
    INSERT INTO facilities
    (name, description)
    VALUES (?, ?)
    """, (name, description))

    conn.commit()

    print("Facility added successfully!")


def view_facilities():
    print("\n------ CAMPUS FACILITIES ------")

    cursor.execute("SELECT * FROM facilities")
    facilities = cursor.fetchall()

    if not facilities:
        print("No facilities found.")
        return

    for facility in facilities:
        print("--------------------------------")
        print("Facility ID :", facility[0])
        print("Name        :", facility[1])
        print("Description :", facility[2])


def facility_management():
    while True:
        print("\n================================")
        print("       CAMPUS FACILITIES")
        print("================================")
        print("1. Add Facility")
        print("2. View Facilities")
        print("3. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_facility()

        elif choice == "2":
            view_facilities()

        elif choice == "3":
            break

        else:
            print("Invalid choice.")


# ------------------------------------------------------------
# EVENT MANAGEMENT
# ------------------------------------------------------------

def add_event():
    print("\n------ ADD CAMPUS EVENT ------")

    event_name = input("Enter event name: ")
    date = input("Enter event date: ")
    venue = input("Enter venue: ")
    description = input("Enter event description: ")

    cursor.execute("""
    INSERT INTO events
    (event_name, date, venue, description)
    VALUES (?, ?, ?, ?)
    """, (event_name, date, venue, description))

    conn.commit()

    print("Event added successfully!")


def view_events():
    print("\n------ CAMPUS EVENTS ------")

    cursor.execute("SELECT * FROM events")
    events = cursor.fetchall()

    if not events:
        print("No events available.")
        return

    for event in events:
        print("--------------------------------")
        print("Event ID    :", event[0])
        print("Event Name  :", event[1])
        print("Date        :", event[2])
        print("Venue       :", event[3])
        print("Description :", event[4])


def event_management():
    while True:
        print("\n================================")
        print("        EVENT MANAGEMENT")
        print("================================")
        print("1. Add Event")
        print("2. View Events")
        print("3. Back")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_event()

        elif choice == "2":
            view_events()

        elif choice == "3":
            break

        else:
            print("Invalid choice.")


# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------

def main_menu():

    while True:

        print("\n")
        print("================================================")
        print(" STUDENT & CAMPUS UTILITY MANAGEMENT SYSTEM")
        print("================================================")
        print("1. Student Management")
        print("2. Attendance Management")
        print("3. Campus Notices")
        print("4. Complaint Management")
        print("5. Campus Facilities")
        print("6. Event Management")
        print("7. Exit")
        print("================================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            student_management()

        elif choice == "2":
            attendance_management()

        elif choice == "3":
            notice_management()

        elif choice == "4":
            complaint_management()

        elif choice == "5":
            facility_management()

        elif choice == "6":
            event_management()

        elif choice == "7":
            print("\nThank you for using the system!")
            break

        else:
            print("Invalid choice. Please try again.")


# ------------------------------------------------------------
# START PROGRAM
# ------------------------------------------------------------

main_menu()

# Close database connection
conn.close()
```

```python
# ============================================================
# UNIVERSITY MANAGEMENT SYSTEM
# USING OBJECT-ORIENTED PYTHON
# ============================================================

# ------------------------------------------------------------
# BASE CLASS - PERSON
# ------------------------------------------------------------

class Person:

    def __init__(self, person_id, name, email, phone):
        self._person_id = person_id
        self._name = name
        self._email = email
        self._phone = phone

    def display_details(self):
        print("ID    :", self._person_id)
        print("Name  :", self._name)
        print("Email :", self._email)
        print("Phone :", self._phone)


# ------------------------------------------------------------
# STUDENT CLASS
# INHERITS FROM PERSON
# ------------------------------------------------------------

class Student(Person):

    def __init__(self, student_id, name, email, phone,
                 department, year):
        super().__init__(student_id, name, email, phone)

        self.department = department
        self.year = year
        self.courses = []
        self.marks = {}

    def enroll_course(self, course):
        if course not in self.courses:
            self.courses.append(course)
            print("Course enrolled successfully!")
        else:
            print("Student is already enrolled in this course.")

    def add_marks(self, course_code, marks):
        self.marks[course_code] = marks
        print("Marks added successfully!")

    def calculate_grade(self, marks):

        if marks >= 90:
            return "A+"
        elif marks >= 80:
            return "A"
        elif marks >= 70:
            return "B"
        elif marks >= 60:
            return "C"
        elif marks >= 50:
            return "D"
        else:
            return "F"

    def display_details(self):

        print("\n--------------------------------")
        print("STUDENT DETAILS")
        print("--------------------------------")

        super().display_details()

        print("Department :", self.department)
        print("Year       :", self.year)

        if self.courses:
            print("Courses    :")

            for course in self.courses:
                print("  -", course.course_code,
                      course.course_name)

        if self.marks:
            print("Results    :")

            for course_code, marks in self.marks.items():
                print(
                    "  ",
                    course_code,
                    ":",
                    marks,
                    "Grade:",
                    self.calculate_grade(marks)
                )


# ------------------------------------------------------------
# FACULTY CLASS
# INHERITS FROM PERSON
# ------------------------------------------------------------

class Faculty(Person):

    def __init__(self, faculty_id, name, email, phone,
                 department, designation):

        super().__init__(faculty_id, name, email, phone)

        self.department = department
        self.designation = designation
        self.courses = []

    def assign_course(self, course):

        if course not in self.courses:
            self.courses.append(course)
            print("Course assigned successfully!")
        else:
            print("Course already assigned.")

    def display_details(self):

        print("\n--------------------------------")
        print("FACULTY DETAILS")
        print("--------------------------------")

        super().display_details()

        print("Department :", self.department)
        print("Designation:", self.designation)

        if self.courses:
            print("Courses    :")

            for course in self.courses:
                print("  -", course.course_code,
                      course.course_name)


# ------------------------------------------------------------
# COURSE CLASS
# ------------------------------------------------------------

class Course:

    def __init__(self, course_code, course_name,
                 credits, department):

        self.course_code = course_code
        self.course_name = course_name
        self.credits = credits
        self.department = department

    def display_course(self):

        print("--------------------------------")
        print("Course Code :", self.course_code)
        print("Course Name :", self.course_name)
        print("Credits     :", self.credits)
        print("Department  :", self.department)


# ------------------------------------------------------------
# DEPARTMENT CLASS
# ------------------------------------------------------------

class Department:

    def __init__(self, department_id, name, hod):

        self.department_id = department_id
        self.name = name
        self.hod = hod
        self.courses = []

    def add_course(self, course):

        if course not in self.courses:
            self.courses.append(course)

    def display_department(self):

        print("--------------------------------")
        print("Department ID :", self.department_id)
        print("Department    :", self.name)
        print("HOD           :", self.hod)

        if self.courses:

            print("Courses:")

            for course in self.courses:
                print("  -",
                      course.course_code,
                      course.course_name)


# ------------------------------------------------------------
# UNIVERSITY MANAGEMENT CLASS
# ------------------------------------------------------------

class University:

    def __init__(self):

        self.students = []
        self.faculty = []
        self.courses = []
        self.departments = []

    # --------------------------------------------------------
    # STUDENT METHODS
    # --------------------------------------------------------

    def add_student(self):

        print("\n========== ADD STUDENT ==========")

        student_id = input("Enter student ID: ")
        name = input("Enter student name: ")
        email = input("Enter email: ")
        phone = input("Enter phone: ")
        department = input("Enter department: ")
        year = input("Enter year: ")

        student = Student(
            student_id,
            name,
            email,
            phone,
            department,
            year
        )

        self.students.append(student)

        print("\nStudent added successfully!")

    def view_students(self):

        print("\n========== STUDENT LIST ==========")

        if not self.students:
            print("No students available.")
            return

        for student in self.students:
            student.display_details()

    def search_student(self):

        print("\n========== SEARCH STUDENT ==========")

        student_id = input("Enter student ID: ")

        for student in self.students:

            if student._person_id == student_id:
                student.display_details()
                return

        print("Student not found.")

    # --------------------------------------------------------
    # FACULTY METHODS
    # --------------------------------------------------------

    def add_faculty(self):

        print("\n========== ADD FACULTY ==========")

        faculty_id = input("Enter faculty ID: ")
        name = input("Enter faculty name: ")
        email = input("Enter email: ")
        phone = input("Enter phone: ")
        department = input("Enter department: ")
        designation = input("Enter designation: ")

        faculty = Faculty(
            faculty_id,
            name,
            email,
            phone,
            department,
            designation
        )

        self.faculty.append(faculty)

        print("\nFaculty added successfully!")

    def view_faculty(self):

        print("\n========== FACULTY LIST ==========")

        if not self.faculty:
            print("No faculty available.")
            return

        for faculty in self.faculty:
            faculty.display_details()

    # --------------------------------------------------------
    # COURSE METHODS
    # --------------------------------------------------------

    def add_course(self):

        print("\n========== ADD COURSE ==========")

        course_code = input("Enter course code: ")
        course_name = input("Enter course name: ")
        credits = input("Enter credits: ")
        department = input("Enter department: ")

        course = Course(
            course_code,
            course_name,
            credits,
            department
        )

        self.courses.append(course)

        print("\nCourse added successfully!")

    def view_courses(self):

        print("\n========== COURSE LIST ==========")

        if not self.courses:
            print("No courses available.")
            return

        for course in self.courses:
            course.display_course()

    # --------------------------------------------------------
    # DEPARTMENT METHODS
    # --------------------------------------------------------

    def add_department(self):

        print("\n========== ADD DEPARTMENT ==========")

        department_id = input("Enter department ID: ")
        name = input("Enter department name: ")
        hod = input("Enter HOD name: ")

        department = Department(
            department_id,
            name,
            hod
        )

        self.departments.append(department)

        print("\nDepartment added successfully!")

    def view_departments(self):

        print("\n========== DEPARTMENT LIST ==========")

        if not self.departments:
            print("No departments available.")
            return

        for department in self.departments:
            department.display_department()

    # --------------------------------------------------------
    # ENROLL STUDENT IN COURSE
    # --------------------------------------------------------

    def enroll_student(self):

        print("\n========== COURSE ENROLLMENT ==========")

        student_id = input("Enter student ID: ")
        course_code = input("Enter course code: ")

        selected_student = None
        selected_course = None

        for student in self.students:

            if student._person_id == student_id:
                selected_student = student

        for course in self.courses:

            if course.course_code == course_code:
                selected_course = course

        if selected_student is None:
            print("Student not found.")
            return

        if selected_course is None:
            print("Course not found.")
            return

        selected_student.enroll_course(selected_course)

    # --------------------------------------------------------
    # ASSIGN COURSE TO FACULTY
    # --------------------------------------------------------

    def assign_course_to_faculty(self):

        print("\n========== FACULTY COURSE ASSIGNMENT ==========")

        faculty_id = input("Enter faculty ID: ")
        course_code = input("Enter course code: ")

        selected_faculty = None
        selected_course = None

        for faculty in self.faculty:

            if faculty._person_id == faculty_id:
                selected_faculty = faculty

        for course in self.courses:

            if course.course_code == course_code:
                selected_course = course

        if selected_faculty is None:
            print("Faculty not found.")
            return

        if selected_course is None:
            print("Course not found.")
            return

        selected_faculty.assign_course(selected_course)

    # --------------------------------------------------------
    # ADD MARKS
    # --------------------------------------------------------

    def add_marks(self):

        print("\n========== ADD MARKS ==========")

        student_id = input("Enter student ID: ")
        course_code = input("Enter course code: ")

        selected_student = None

        for student in self.students:

            if student._person_id == student_id:
                selected_student = student

        if selected_student is None:
            print("Student not found.")
            return

        enrolled = False

        for course in selected_student.courses:

            if course.course_code == course_code:
                enrolled = True

        if not enrolled:
            print("Student is not enrolled in this course.")
            return

        try:

            marks = float(
                input("Enter marks (0-100): ")
            )

            if marks < 0 or marks > 100:
                print("Marks must be between 0 and 100.")
                return

            selected_student.add_marks(
                course_code,
                marks
            )

        except ValueError:

            print("Please enter a valid number.")

    # --------------------------------------------------------
    # DISPLAY RESULT
    # --------------------------------------------------------

    def display_result(self):

        print("\n========== STUDENT RESULT ==========")

        student_id = input("Enter student ID: ")

        for student in self.students:

            if student._person_id == student_id:

                print("\nStudent Name:", student._name)
                print("Department :", student.department)

                if not student.marks:
                    print("No marks available.")
                    return

                print("\n--------------------------------")
                print("COURSE      MARKS      GRADE")
                print("--------------------------------")

                for course_code, marks in student.marks.items():

                    grade = student.calculate_grade(marks)

                    print(
                        course_code,
                        "      ",
                        marks,
                        "      ",
                        grade
                    )

                return

        print("Student not found.")


# ============================================================
# MAIN PROGRAM
# ============================================================

university = University()


# ------------------------------------------------------------
# MAIN MENU
# ------------------------------------------------------------

def main_menu():

    while True:

        print("\n")
        print("================================================")
        print("       UNIVERSITY MANAGEMENT SYSTEM")
        print("       USING OBJECT-ORIENTED PYTHON")
        print("================================================")

        print("1. Student Management")
        print("2. Faculty Management")
        print("3. Course Management")
        print("4. Department Management")
        print("5. Enroll Student in Course")
        print("6. Assign Course to Faculty")
        print("7. Add Student Marks")
        print("8. Display Student Result")
        print("9. Exit")

        print("================================================")

        choice = input("Enter your choice: ")

        # Student Management
        if choice == "1":

            while True:

                print("\n---------- STUDENT MANAGEMENT ----------")

                print("1. Add Student")
                print("2. View Students")
                print("3. Search Student")
                print("4. Back")

                student_choice = input(
                    "Enter your choice: "
                )

                if student_choice == "1":
                    university.add_student()

                elif student_choice == "2":
                    university.view_students()

                elif student_choice == "3":
                    university.search_student()

                elif student_choice == "4":
                    break

                else:
                    print("Invalid choice.")

        # Faculty Management
        elif choice == "2":

            while True:

                print("\n---------- FACULTY MANAGEMENT ----------")

                print("1. Add Faculty")
                print("2. View Faculty")
                print("3. Back")

                faculty_choice = input(
                    "Enter your choice: "
                )

                if faculty_choice == "1":
                    university.add_faculty()

                elif faculty_choice == "2":
                    university.view_faculty()

                elif faculty_choice == "3":
                    break

                else:
                    print("Invalid choice.")

        # Course Management
        elif choice == "3":

            while True:

                print("\n---------- COURSE MANAGEMENT ----------")

                print("1. Add Course")
                print("2. View Courses")
                print("3. Back")

                course_choice = input(
                    "Enter your choice: "
                )

                if course_choice == "1":
                    university.add_course()

                elif course_choice == "2":
                    university.view_courses()

                elif course_choice == "3":
                    break

                else:
                    print("Invalid choice.")

        # Department Management
        elif choice == "4":

            while True:

                print("\n---------- DEPARTMENT MANAGEMENT ----------")

                print("1. Add Department")
                print("2. View Departments")
                print("3. Back")

                department_choice = input(
                    "Enter your choice: "
                )

                if department_choice == "1":
                    university.add_department()

                elif department_choice == "2":
                    university.view_departments()

                elif department_choice == "3":
                    break

                else:
                    print("Invalid choice.")

        # Enrollment
        elif choice == "5":

            university.enroll_student()

        # Faculty course assignment
        elif choice == "6":

            university.assign_course_to_faculty()

        # Marks
        elif choice == "7":

            university.add_marks()

        # Result
        elif choice == "8":

            university.display_result()

        # Exit
        elif choice == "9":

            print("\nThank you for using")
            print("University Management System!")

            break

        else:

            print("Invalid choice. Please try again.")


# ------------------------------------------------------------
# START APPLICATION
# ------------------------------------------------------------

main_menu()
```

```python
# ================================================================
# STUDENT MANAGEMENT PACKAGE & DEPENDENCY MANAGEMENT SYSTEM
# ONE-CELL GOOGLE COLAB VERSION
# ================================================================

# ------------------------------------------------
# 1. INSTALL DEPENDENCY
# ------------------------------------------------

!pip install -q tabulate


# ------------------------------------------------
# 2. IMPORT LIBRARIES
# ------------------------------------------------

from tabulate import tabulate


# ================================================================
# STUDENT CLASS
# ================================================================

class Student:

    def __init__(self, student_id, name, department, year, email):
        self.student_id = student_id
        self.name = name
        self.department = department
        self.year = year
        self.email = email
        self.marks = {}

    def display_student(self):

        print("\n----------------------------------------")
        print("          STUDENT DETAILS")
        print("----------------------------------------")

        print("Student ID :", self.student_id)
        print("Name       :", self.name)
        print("Department :", self.department)
        print("Year       :", self.year)
        print("Email      :", self.email)

    def add_marks(self, subject, marks):

        if marks < 0 or marks > 100:
            print("Marks must be between 0 and 100.")
            return

        self.marks[subject] = marks

        print("Marks added successfully!")

    def calculate_total(self):

        return sum(self.marks.values())

    def calculate_average(self):

        if len(self.marks) == 0:
            return 0

        return self.calculate_total() / len(self.marks)

    def calculate_grade(self):

        average = self.calculate_average()

        if average >= 90:
            return "A+"
        elif average >= 80:
            return "A"
        elif average >= 70:
            return "B"
        elif average >= 60:
            return "C"
        elif average >= 50:
            return "D"
        else:
            return "F"

    def display_marks(self):

        print("\n----------------------------------------")
        print("             MARKS DETAILS")
        print("----------------------------------------")

        if not self.marks:
            print("No marks available.")
            return

        table = []

        for subject, marks in self.marks.items():

            table.append([
                subject,
                marks
            ])

        print(
            tabulate(
                table,
                headers=["Subject", "Marks"],
                tablefmt="grid"
            )
        )

        print("Total   :", self.calculate_total())
        print("Average :", round(self.calculate_average(), 2))
        print("Grade   :", self.calculate_grade())


# ================================================================
# STUDENT MANAGEMENT SYSTEM
# ================================================================

class StudentManagementSystem:

    def __init__(self):

        self.students = []

    # ------------------------------------------------------------
    # ADD STUDENT
    # ------------------------------------------------------------

    def add_student(self):

        print("\n========== ADD STUDENT ==========")

        student_id = input("Enter student ID: ")

        # Check duplicate ID
        for student in self.students:

            if student.student_id == student_id:

                print("Student ID already exists!")
                return

        name = input("Enter student name: ")
        department = input("Enter department: ")
        year = input("Enter year: ")
        email = input("Enter email: ")

        student = Student(
            student_id,
            name,
            department,
            year,
            email
        )

        self.students.append(student)

        print("\nStudent added successfully!")

    # ------------------------------------------------------------
    # VIEW STUDENTS
    # ------------------------------------------------------------

    def view_students(self):

        print("\n========== STUDENT LIST ==========")

        if not self.students:

            print("No students available.")
            return

        table = []

        for student in self.students:

            table.append([
                student.student_id,
                student.name,
                student.department,
                student.year,
                student.email
            ])

        print(
            tabulate(
                table,
                headers=[
                    "ID",
                    "Name",
                    "Department",
                    "Year",
                    "Email"
                ],
                tablefmt="grid"
            )
        )

    # ------------------------------------------------------------
    # SEARCH STUDENT
    # ------------------------------------------------------------

    def search_student(self):

        print("\n========== SEARCH STUDENT ==========")

        student_id = input("Enter student ID: ")

        for student in self.students:

            if student.student_id == student_id:

                student.display_student()

                return

        print("Student not found.")

    # ------------------------------------------------------------
    # ADD MARKS
    # ------------------------------------------------------------

    def add_marks(self):

        print("\n========== ADD MARKS ==========")

        student_id = input("Enter student ID: ")

        selected_student = None

        for student in self.students:

            if student.student_id == student_id:

                selected_student = student
                break

        if selected_student is None:

            print("Student not found.")
            return

        subject = input("Enter subject name: ")

        try:

            marks = float(
                input("Enter marks (0-100): ")
            )

            selected_student.add_marks(
                subject,
                marks
            )

        except ValueError:

            print("Please enter a valid number.")

    # ------------------------------------------------------------
    # VIEW MARKS
    # ------------------------------------------------------------

    def view_marks(self):

        print("\n========== VIEW MARKS ==========")

        student_id = input("Enter student ID: ")

        for student in self.students:

            if student.student_id == student_id:

                student.display_marks()

                return

        print("Student not found.")

    # ------------------------------------------------------------
    # GENERATE REPORT
    # ------------------------------------------------------------

    def generate_report(self):

        print("\n========== STUDENT REPORT ==========")

        student_id = input("Enter student ID: ")

        for student in self.students:

            if student.student_id == student_id:

                print("\n")
                print("================================================")
                print("             STUDENT REPORT")
                print("================================================")

                print("Student ID :", student.student_id)
                print("Name       :", student.name)
                print("Department :", student.department)
                print("Year       :", student.year)
                print("Email      :", student.email)

                print("------------------------------------------------")

                if not student.marks:

                    print("No marks available.")

                else:

                    table = []

                    for subject, marks in student.marks.items():

                        table.append([
                            subject,
                            marks
                        ])

                    print(
                        tabulate(
                            table,
                            headers=["Subject", "Marks"],
                            tablefmt="grid"
                        )
                    )

                    print(
                        "Total   :",
                        student.calculate_total()
                    )

                    print(
                        "Average :",
                        round(
                            student.calculate_average(),
                            2
                        )
                    )

                    print(
                        "Grade   :",
                        student.calculate_grade()
                    )

                print("================================================")

                return

        print("Student not found.")

    # ------------------------------------------------------------
    # DELETE STUDENT
    # ------------------------------------------------------------

    def delete_student(self):

        print("\n========== DELETE STUDENT ==========")

        student_id = input("Enter student ID: ")

        for student in self.students:

            if student.student_id == student_id:

                self.students.remove(student)

                print("Student deleted successfully!")

                return

        print("Student not found.")


# ================================================================
# MAIN MENU
# ================================================================

def main():

    system = StudentManagementSystem()

    while True:

        print("\n")
        print("========================================================")
        print("       STUDENT MANAGEMENT PACKAGE &")
        print("       DEPENDENCY MANAGEMENT SYSTEM")
        print("========================================================")

        print("1. Add Student")
        print("2. View Students")
        print("3. Search Student")
        print("4. Add Marks")
        print("5. View Marks")
        print("6. Generate Student Report")
        print("7. Delete Student")
        print("8. Exit")

        print("========================================================")

        choice = input("Enter your choice: ")

        if choice == "1":

            system.add_student()

        elif choice == "2":

            system.view_students()

        elif choice == "3":

            system.search_student()

        elif choice == "4":

            system.add_marks()

        elif choice == "5":

            system.view_marks()

        elif choice == "6":

            system.generate_report()

        elif choice == "7":

            system.delete_student()

        elif choice == "8":

            print("\nThank you for using the system!")
            print("Project completed successfully.")

            break

        else:

            print("Invalid choice. Please try again.")


# ================================================================
# START PROJECT
# ================================================================

main()
```
